# 🟡 Solution: MLP Forward Pass (NumPy)

```python
import numpy as np
```

```python
# ✅ SOLUTION

def mlp_forward(X, params):
    # X: (N, d_in);  params: list of (W, b), W: (in, out), b: (out,)
    A = X
    last = len(params) - 1

    for i, (W, b) in enumerate(params):
        Z = A @ W + b                      # affine map, b broadcasts over rows
        A = np.maximum(Z, 0) if i < last else Z   # ReLU on hidden layers only

    return A
```

```python
# Verify
np.random.seed(0)
X = np.random.randn(5, 4)
params = [(np.random.randn(4, 6), np.random.randn(6)),
          (np.random.randn(6, 3), np.random.randn(3))]

out = mlp_forward(X, params)
print("output shape:", out.shape)
print("one layer == affine:", np.allclose(mlp_forward(X, [params[0]]), X @ params[0][0] + params[0][1]))
```
