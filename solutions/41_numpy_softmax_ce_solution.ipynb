# 🟡 Solution: Softmax Cross-Entropy + Gradient (NumPy)

```python
import numpy as np
```

```python
# ✅ SOLUTION

def softmax_cross_entropy(logits, labels):
    # logits: (N, C);  labels: (N,) int
    N = logits.shape[0]
    rows = np.arange(N)

    # Stable softmax: shift by the row max so the largest exponent is e^0 = 1
    z = logits - logits.max(axis=1, keepdims=True)
    exp_z = np.exp(z)
    p = exp_z / exp_z.sum(axis=1, keepdims=True)

    # Mean negative log-likelihood of the true class
    loss = float(-np.log(p[rows, labels]).mean())

    # The fused gradient: (predicted - actual) / N
    dlogits = p.copy()
    dlogits[rows, labels] -= 1.0
    dlogits /= N

    return loss, dlogits
```

```python
# Verify
np.random.seed(0)
logits = np.random.randn(6, 4)
labels = np.array([0, 1, 2, 3, 0, 1])

loss, dlogits = softmax_cross_entropy(logits, labels)
print("loss        :", loss)
print("rows sum ~0 :", np.allclose(dlogits.sum(axis=1), 0))

uniform_loss, _ = softmax_cross_entropy(np.zeros((4, 10)), np.array([0, 3, 7, 9]))
print("uniform loss:", uniform_loss, "== log(10) =", np.log(10))

big, _ = softmax_cross_entropy(np.array([[1000.0, 1001.0, 999.0]]), np.array([1]))
print("stable      :", np.isfinite(big))
```
