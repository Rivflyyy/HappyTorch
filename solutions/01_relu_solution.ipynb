# 🟢 Solution: Implement ReLU

Reference solution for the ReLU activation function.

$$\text{ReLU}(x) = \max(0, x)$$

### Exercise contract
At `x=0`, ReLU is not differentiable; any finite subgradient in `[0, 1]` is accepted. Gradients at negative/positive inputs must be 0/1. There is no hardware-dependent time limit in the correctness tests.

```python
import torch
```

```python
# ✅ SOLUTION

def relu(x: torch.Tensor) -> torch.Tensor:
    return torch.where(x > 0, x, torch.zeros_like(x))
```

```python
# Verify
x = torch.tensor([-2., -1., 0., 1., 2.])
print("Input: ", x)
print("Output:", relu(x))
```
