### Core Idea

Every neural network is a chain of two ingredients: an **affine map** and a **nonlinearity**.

$$A^{(0)} = X, \qquad Z^{(l)} = A^{(l-1)} W^{(l)} + b^{(l)}, \qquad A^{(l)} = \mathrm{ReLU}\!\left(Z^{(l)}\right)$$

Two details carry all the meaning:

1. **The last layer stays linear.** It emits *logits* — unbounded real scores. Squashing them with
   ReLU would clip half the score range and destroy the softmax that follows.
2. **Without the nonlinearity the depth is fake.** $XW_1W_2 = X(W_1W_2)$ — a stack of affine maps is
   just one affine map. ReLU is the cheapest thing that breaks that collapse.

Shapes are the whole game. Here $W$ has shape `(in, out)` and rows of $X$ are samples, so it is
`A @ W` (not `A @ W.T` like `nn.Linear`). The bias `(out,)` broadcasts down every row.

Keep this function clean — the [backprop problem](#) reuses exactly this forward pass, plus a cache.
