# 🔴 Hard: KV Cache for Attention

Implement **KV Cache** for efficient LLM inference during generation.

KV Cache stores past Key and Value tensors to avoid recomputation:
- `append(k, v)`: Add new K, V to cache
- `get()`: Return all cached K, V
- `clear()`: Reset cache

### Signature
```python
class KVCache:
    def __init__(self, num_heads: int, head_dim: int, max_seq_len: int, batch_size: int):
        ...
    def append(self, k: torch.Tensor, v: torch.Tensor) -> None:
        ...
    def get(self) -> tuple[torch.Tensor, torch.Tensor]:
        ...
    def clear(self) -> None:
        ...
```

### Current grader interface and conventions
Expose **`self.k_cache`**, **`self.v_cache`**, and **`self.seq_len`**; the current judge checks these exact names. `seq_len` is an integer attribute/property giving the number of stored tokens, initially 0 and reset to 0 by `clear()`.

`append(k, v)` adds tokens along dimension **2** of `(batch_size, num_heads, new_seq_len, head_dim)`. `get()` returns the stored prefix of both caches in arrival order, with shape `(batch_size, num_heads, seq_len, head_dim)`. Preallocated tensors may have capacity `max_seq_len`, but unused capacity must not appear in the returned prefix. The examples/tests use CPU float32 and appends within capacity.
