# 🔴 Hard: AdaLN-Zero

Implement **AdaLN-Zero**, a variant with zero initialization for stable training.

Like AdaLN but with zero initialization so the residual branch initially outputs zero:
- $\gamma = \beta = 0$ at initialization
- Includes a gate parameter for residual connections

### Signature
```python
class AdaLNZero(nn.Module):
    def __init__(self, hidden_dim: int, cond_dim: int):
        ...
```

### Requirements
- Final output projection must be zero-initialized
- Include a `gate_proj` for residual gate
- Output at init should be close to zero

### Current grader interface and conventions
Return the residual **branch** of shape `(B, S, hidden_dim)` for x `(B, S, hidden_dim)` and cond `(B, cond_dim)`. A zero branch output means that a caller computing `x + branch` starts as identity; it does not mean this layer returns x.

Expose **`self.gate_proj = nn.Linear(cond_dim, hidden_dim)`** with zero-initialized weight and bias, and use it to produce the per-sample gate. The judge accesses `gate_proj.weight` and `gate_proj.bias` directly. Broadcast the gate across tokens. It also requires gradients through x/cond and a nonzero output after a few optimization steps; merely returning zero or adding an unused module with the right name does not implement the intended behavior.
