# 🔴 Hard: AdaLN (Adaptive Layer Normalization)

Implement **AdaLN**, used in DiT and Diffusion Transformers for conditional generation.

AdaLN modulates LayerNorm using a conditioning signal:
$$\text{AdaLN}(x, c) = \gamma(c) \odot \text{LN}(x) + \beta(c)$$

Where $\gamma$ and $\beta$ are predicted from the condition $c$.

### Signature
```python
class AdaLN(nn.Module):
    def __init__(self, hidden_dim: int, cond_dim: int):
        ...
    def forward(self, x: torch.Tensor, cond: torch.Tensor) -> torch.Tensor:
        ...
```

### Requirements
- Normalize over the last dimension with population variance, `eps=1e-5`, and no fixed affine parameters (`nn.LayerNorm(..., elementwise_affine=False)` is one option)
- Use learned projections or a small MLP to predict $\gamma$ and $\beta$ from condition
- Shape: x is (batch, seq, hidden_dim), cond is (batch, cond_dim)

Each sample gets its own scale/shift vectors, shared across its tokens. Support autograd through x, condition, and the conditioning parameters.

Internal names are unrestricted: separate projections and a fused projection plus `chunk` are both valid. The judge exercises nonzero parameters and compares behavior; it does not require `gamma_fc`, `beta_fc`, `norm`, or `cond_mlp`.
