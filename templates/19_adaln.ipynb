# 🔴 Hard: AdaLN (Adaptive Layer Normalization)

Implement **AdaLN**, used in DiT and Diffusion Transformers for conditional generation.

AdaLN modulates LayerNorm using a conditioning signal:
$$\text{AdaLN}(x, c) = \gamma(c) \odot \text{LN}(x) + \beta(c)$$

Where $\gamma$ and $\beta$ are predicted from the condition $c$.

### Signature
```python
class AdaLN(nn.Module):
    def __init__(self, hidden_dim: int, cond_dim: int):
        ...
    def forward(self, x: torch.Tensor, cond: torch.Tensor) -> torch.Tensor:
        ...
```

### Requirements
- Use `nn.LayerNorm` for the base normalization
- Use a small MLP (SiLU + Linear) to predict $\gamma$ and $\beta$ from condition
- Shape: x is (batch, seq, hidden_dim), cond is (batch, cond_dim)

### Current grader interface and conventions
Expose **`self.norm`** as `nn.LayerNorm(hidden_dim, elementwise_affine=False)` and **`self.cond_mlp`** as the conditioning module, e.g. `nn.Sequential(nn.SiLU(), nn.Linear(cond_dim, 2 * hidden_dim))`. The current tests inspect these exact names and zero the parameters of `cond_mlp` in one check. Renaming them can fail even when the computation is equivalent.

Split the last conditioning dimension into gamma and beta, each `(B, hidden_dim)`, then broadcast across the sequence dimension. Normalize over the last feature dimension with population variance and `eps=1e-5`. Use `gamma * LN(x) + beta` and preserve autograd. In this formula, zeroing both predictions gives zero output, not plain LayerNorm; the test titled “Condition-independent when gamma=1, beta=0” only checks output shape after zeroing the module.
