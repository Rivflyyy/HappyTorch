# 🔴 Hard: LoRA (Low-Rank Adaptation)

Implement **LoRA** for parameter-efficient fine-tuning of large models.

LoRA adds a low-rank decomposition to existing weights:
$$W' = W + \Delta W = W + \frac{\alpha}{r}BA$$

Where $B \in \mathbb{R}^{d_{out} \times r}$ and $A \in \mathbb{R}^{r \times d_{in}}$.

### Signature
```python
class LoRALinear(nn.Module):
    def __init__(self, in_features: int, out_features: int, rank: int, alpha: float = 1.0):
        ...
    def forward(self, x: torch.Tensor) -> torch.Tensor:
        ...
```

### Requirements
- Initialize B with zeros (so the base layer is initially unchanged)
- Initialize A with Kaiming uniform or normal
- Forward: `x @ W.T + (x @ A.T @ B.T) * alpha / rank`

### Current grader interface and conventions
Expose `self.W` `(out_features, in_features)`, `self.A` `(rank, in_features)`, and `self.B` `(out_features, rank)` as `nn.Parameter`s. Keep numeric attributes `self.rank` and `self.alpha`. Tests read these names, set A/B to nonzero values, inspect A/B gradients, and count registered parameters including W. W may be frozen or trainable; A/B must be trainable for positive rank.

The layer is bias-free: input `(..., in_features)` maps to `(..., out_features)`. Support `rank=0` by returning `x @ W.T` without dividing by rank; zero-sized A/B rank dimensions are valid. Zero B means the initial output equals the **base layer**, not necessarily x.
