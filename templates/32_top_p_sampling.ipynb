Implement **top-p (nucleus) sampling**.

Sort tokens by probability, keep the smallest prefix whose cumulative probability reaches `p`, renormalize, then sample from that nucleus.

### Signature
```python
def top_p_sample(logits: torch.Tensor, p: float) -> torch.Tensor:
    ...
```

### Rules
- Support both 1-D and batched logits
- Always keep at least one token
- Sample one token per row

### Example
```
token_ids = top_p_sample(logits, p=0.9)
```

### Current grader interface and conventions
Accept logits `(vocab_size,)` or `(B, vocab_size)` and return integer indices of shape `()` or `(B,)`, with `0 < p <= 1`.

The current fixed-seed check follows the reference **row by row**: sort probabilities descending, filter, place retained probabilities back in **original vocabulary order**, renormalize, and call `torch.multinomial(..., num_samples=1)` separately for each row. A batched call or sampling directly in sorted order can give different seeded tokens even for the same distribution.

There is also a boundary convention in the current reference: it keeps entries with cumulative probability `<= p`, plus the first entry whose cumulative probability is `> p`, and always retains the first token. At exact equality this includes one extra token compared with the usual “smallest prefix reaching p” definition above. This describes the existing reference/check; the mathematical definition and this edge case are not identical.
