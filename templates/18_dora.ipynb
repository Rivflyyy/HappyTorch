# 🔴 Hard: DoRA (Weight-Decomposed LoRA)

Implement **DoRA**, an improved variant of LoRA that decomposes weights into magnitude and direction.

DoRA decomposes $W$ into magnitude $m$ and direction $V$:
$$W = m \cdot \frac{V}{||V||}$$

LoRA updates the direction: $V' = V + (\alpha/r)BA$

Final weight: $W' = m \cdot \frac{V + (\alpha/r)BA}{||V + (\alpha/r)BA||}$

```python

        ...
```

### Requirements
- Initialize $m$ as the L2 norm of $W$ per output dimension
- Initialize $B$ with zeros
- Apply normalization to $V + (\alpha/r)BA$ before multiplying by $m$

### Current grader interface and conventions
Expose `self.W` `(out_features, in_features)`, `self.A` `(rank, in_features)`, `self.B` `(out_features, rank)`, and `self.m` `(out_features,)`, plus numeric `self.rank` and `self.alpha`. Register m/A/B as trainable `nn.Parameter`s. The judge reads these attributes and installs known nonzero weights; a column-shaped m is not the tested interface.

For positive rank, use `direction = W + (alpha / rank) * (B @ A)`. Normalize **each output row** along `dim=1`, then multiply by `m[:, None]`. Initialize m from W's row norms, A randomly, and B to zero. Forward is bias-free: `x @ effective_weight.T`. Preserve autograd through x and m/A/B.
